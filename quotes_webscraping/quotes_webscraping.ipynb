```python
import requests
from bs4 import BeautifulSoup
```

```python
main_url = "https://quotes.toscrape.com/"
response = requests.get(main_url)
soup = BeautifulSoup(response.text, "html.parser")
print("Status code:", response.status_code)
```

```python
quotes_data = []

tags_box = soup.find("div", class_="col-md-4 tags-box")
links = tags_box.find_all("a")  

for each in links:
    quotes_data.append({
        'title': each.get_text(strip=True),
        'url': each.get('href')
    })

quotes_data
```

```python
url=[]
for each in quotes_data:
    url.append(each['url'])

url
```

```python
category_quotes=[]
all_category={}
for each in url:
    tag_name = each.strip("/").split("/")[-1]
    response = requests.get(main_url+each)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("div", class_="quote")
    for quote in quotes:
        text = quote.find("span", class_="text").get_text(strip=True)
        author = quote.find("small", class_="author").get_text(strip=True)
        tags = [t.get_text(strip=True) for t in quote.find_all("a", class_="tag")]

        category_quotes.append({"text":text,"author":author,"tags":tags})

    all_category[tag_name]=category_quotes

all_category
```

```python
import pandas as pd

for tag_name, quotes in all_category.items():
    name = f"df_{tag_name.replace('-','_')}"
    globals()[name] = pd.DataFrame(quotes)
```
